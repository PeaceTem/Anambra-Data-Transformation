```python
import pandas as pd
```

```python
sheets = ["Table 202"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna('')
    dfs[index].replace(['NA', 'NIL', '-'], '', inplace=True)

dfs[0].head()
```

```python
columns = ['media_houses', 'year', 'type', 'category', 'location']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        transformed_data = [{
            'media_houses': row['MEDIA HOUSES'],
            'year': 2021,
            'type': row['TYPE'],
            'category': row['CATEGORY'],
            'location': row['LOCATION/ADDRESS']
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)

new_df.head(22)
```
