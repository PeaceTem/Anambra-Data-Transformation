```python
import pandas as pd
```

```python
years = [2025]
dfs = []
for index, sheet in enumerate(years):
    dfs.append(pd.read_excel("anthrax_vaccination.xlsx"))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head(20)
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga_id', 'year', 'livestock', 'num_of_vaccination']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"].upper().strip(), "lga_id"].values) == 0):
            print(row["LGA"].upper())
        # print(row)
        else:
            lga_id = lga_df.loc[lga_df['lga_name'] == row["LGA"].upper().strip(), "lga_id"].values[0]
            for col in df.columns[2:]:
                transformed_data = [
                    {
                        'lga_id': lga_id,
                        'year': year,
                        'livestock': col,
                        'num_of_vaccination': row[col]
                    }
                ]
                new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['num_of_vaccination']=new_df['num_of_vaccination'].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposedb/anthrax_vaccination_transposed.xlsx", index=False)
print("Done!")
```
