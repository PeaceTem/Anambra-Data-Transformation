```python
import pandas as pd
```

```python
sheets = ["Sheet1", "Sheet2"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("crops_volume_of_production.xlsx", sheet_name=sheets[index]))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['year', 'crop', 'volume']

new_df = pd.DataFrame(columns=columns)

for index, row in dfs[0].iterrows():
    for col in dfs[0].columns[1:]:
        transformed_data = [{
            'year': row['Year'],
            'crop' : col,
            'volume' : row[col]
        }]
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['year']=new_df['year'].astype(int)
new_df.head()
```

```python
new_df.to_excel("crop_volume_of_production_transposed.xlsx", index=False)
print("Done!")
```
