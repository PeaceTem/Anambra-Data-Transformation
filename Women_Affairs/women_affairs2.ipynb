```python
import pandas as pd
```

```python
sheets = ["Sheet1"]
file_names = ["women_affairs.xlsx"]
years = [2024]
dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet1"))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['sn', 'year', 'question', 'response']

new_df = pd.DataFrame(columns=columns)
year = 2024

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        for col in df.columns[1:]:
            if(row[col] in ('0', 0)): continue
            transformed_data = [{
                'sn': row["S/N"],
                'year': year,
                'question': col,
                'response': row[col],
            }]

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(50)
```

```python
new_df.to_excel("Women_affairs_transposed.xlsx", index=False)
print("Done!")
```
