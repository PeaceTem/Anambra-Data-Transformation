```python
import pandas as pd
```

```python
sheets = ["Table 222"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga_id', 'year', 'management', 'gender', 'num_trained_personnel' ]


new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values) == 0):
            print(row["LGA"])
        for col in df.columns[1:]:
            t = col.split('-')
            if 'SANITATION' in t[1]: continue
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'year': t[0].strip(),
                'management': t[1].split()[-2],
                'gender': t[2].strip(),
                'num_trained_personnel': row[col],
            }]

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(50)
```

```python
columns = ['lga_id', 'year', 'management', 'gender', 'num_trained_personnel' ]


sanitation_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values) == 0):
            print(row["LGA"])
        for col in df.columns[1:]:
            t = col.split('-')
            if 'WATER' in t[1]: continue
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'year': t[0].strip(),
                'management': t[1].split()[-2],
                'gender': t[2].strip(),
                'num_trained_personnel': row[col],
            }]

            sanitation_df = pd.concat([sanitation_df, pd.DataFrame(transformed_data)], ignore_index=True)


sanitation_df.head(22)
```

```python
new_df["num_trained_personnel"] = new_df["num_trained_personnel"].astype(int)
sanitation_df["num_trained_personnel"] = sanitation_df["num_trained_personnel"].astype(int)
new_df.head()
```

```python

new_df = pd.concat([new_df, sanitation_df], ignore_index=True)


new_df.to_excel("water_and_sanitation_management_training.xlsx", index=False)
print("Done!")

# sanitation_df.to_excel("sanitation_management_training.xlsx", index=False)
# print("Done!")
```
