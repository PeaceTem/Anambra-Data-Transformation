```python
import pandas as pd
```

```python
sheets = ["Sheet1"]
file_names = ["phase1.xlsx", "phase2.xlsx"]
batches = ["Phase 1", "Phase 2"]
dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet1"))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga', 'year', 'batch', 'gender', 'population']

new_df = pd.DataFrame(columns=columns)
year = 2024

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        if(row["LGA"]==0): continue

        transformed_data = [{
            'lga': row["LGA"],
            'year': year,
            'batch': batches[i],
            'gender': "Male",
            'population': row["No. OF MALE"]
        },{
            'lga': row["LGA"],
            'year': year,
            'batch': batches[i],
            'gender': "Female",
            'population': row["No. OF FEMALE"]
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.tail(22)
```

```python
new_df['population']=new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("youth_population.xlsx", index=False)
print("Done!")
```
