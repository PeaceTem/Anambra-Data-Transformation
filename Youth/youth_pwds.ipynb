```python
import pandas as pd
```

```python
sheets = ["Sheet1"]
file_names = ["job_creation_statistics.xlsx", "phase2.xlsx"]
batches = ["Phase 1", "Phase 2"]
sessions = ["2022/2023", "2024/2025"]

dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet1"))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga', 'session', 'batch', 'pwd']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        if(row["LGA"]==0): continue

        transformed_data = [{
            'lga': row["LGA"],
            'session': sessions[i],
            'batch': batches[i],
            'pwd': row["No. OF PWDs"]
        }]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['pwd']=new_df['pwd'].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposedb/youth_pwds.xlsx", index=False)
print("Done!")
```
