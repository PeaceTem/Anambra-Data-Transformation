```python
import pandas as pd
```

```python
sheets = ["Sheet2"]
file_names = ["job_creation_statistics.xlsx"]
batches = ["Phase 1"]
sessions = ["2022/2023"]

dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet2"))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['session', 'batch', 'job', 'no_of_jobs_created']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        transformed_data = [{
            'session': sessions[i],
            'batch': batches[i],
            'job': row["JOBS"],
            'no_of_jobs_created': row["TOTAL NO. OF JOBS CREATED"],
        }]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['no_of_jobs_created']=new_df['no_of_jobs_created'].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposedb/jobs_created.xlsx", index=False)
print("Done!")
```
