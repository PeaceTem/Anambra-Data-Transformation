```python
import pandas as pd
```

```python
years = [2019, 2020, 2021]
sheets = ["Table 189"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['constituency', 'assembly', 'name', 'gender', 'position', 'date_of_birth', 'political_party', 'first_election_year_into_the_house', 'academic_qualification', 'phone_number', 'email']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        transformed_data = [{
            'constituency': row['CONSTITUENCY'].replace("\n", " "),
            'assembly': '7th',
            'name': row["NAME"].replace("\n", " "),
            'gender': "Male" if row["GENDER"] == 'M' else "Female",
            'position': row["POSITION"].replace("\n", " "),
            'date_of_birth': row["DATE OF BIRTH"].strftime("%m-%d-%Y"),
            'political_party': row["POLITICAL PARTY"].replace("\n", " "),
            'first_election_year_into_the_house': row["FIRST ELECTION YEAR INTO THE HOUSE"].replace("\n", " "),
            'academic_qualification': row["ACADEMIC QUALIFICATION"].replace("\n", " "),
            'phone_number': '0'+str(row["PHONE NUMBER"]),
            'email': row["EMAIL"].replace("\n", " "),
        }]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df.to_excel("legislators_profile.xlsx", index=False)
print("Done!")
```
