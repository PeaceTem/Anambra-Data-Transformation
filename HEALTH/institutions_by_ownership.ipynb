```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 46", "Table 47", "Table 48", "Table 49"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[1].head()
```

```python
columns = ['lga_id', 'ownership', 'year', 'population']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values) == 0):
            print(row["LGA"])
        # print(row)
        else:
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'ownership': "FEDERAL GOVERNMENT OWNED",
                'year': year,
                'population': row["FEDERAL GOVERNMENT OWNED"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'ownership': "STATE GOVERNMENT TEACHING HOSPITAL",
                'year': year,
                'population': row["STATE GOVERNMENT TEACHING HOSPITAL"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'ownership': "STATE GOVERNMENT OWNED (GENERAL HOSPITAL)",
                'year': year,
                'population': row["STATE GOVERNMENT OWNED (GENERAL HOSPITAL)"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'ownership': "LOCAL GOVERNMENT OWNED (PHC)",
                'year': year,
                'population': row["LOCAL GOVERNMENT OWNED (PHC)"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'ownership': "PRIVATE OWNED HOSPITAL",
                'year': year,
                'population': row["PRIVATE OWNED HOSPITAL"],
            },]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['population']=new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("institutions_by_ownership.xlsx", index=False)
print("Done!")
```
