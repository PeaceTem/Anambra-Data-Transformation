```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 70", "Table 71", "Table 72", "Table 73"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[1].head()
```

```python
columns = ['year', 'cadre', 'gender', 'hospital_type', 'workforce']

new_df = pd.DataFrame(columns=columns)
ref = {
    'PUBLIC MALE' : {
        'gender': 'Male',
        'hospital_type': 'Public',
        },
    'PUBLIC FEMALE' : {
        'gender': 'Female',
        'hospital_type': 'Public',
        },
    'NAUTH MALE' : {
        'gender': 'Male',
        'hospital_type': 'Nauth',
        },
    'NAUTH FEMALE' : {
        'gender': 'Female',
        'hospital_type': 'Nauth',
        },
    'PRIVATE MALE' : {
        'gender': 'Male',
        'hospital_type': 'Private',
        },
    'PRIVATE FEMALE' : {
        'gender': 'Female',
        'hospital_type': 'Private',
        },
}

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        for col in df.columns[1:]:
            transformed_data = [{
                'year': year,
                'cadre': row['CADRE'],
                'gender': ref[col]['gender'],
                'hospital_type': ref[col]['hospital_type'],
                'workforce': row[col],
            }]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
print("len", len(new_df['cadre'].unique()))
new_df['cadre'].unique()
```

```python
new_df['workforce']=new_df['workforce'].astype(int)
new_df['year']=new_df['year'].astype(int)

new_df.head()
```

```python
new_df.to_excel("transposed/health_workforce_by_cadre.xlsx", index=False)
print("Done!")
```
