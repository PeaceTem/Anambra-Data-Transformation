```python
import pandas as pd
```

```python
# %pip install xlrd
# years = [2021, 2022, 2023, 2024, 2025]
dfs=[]
file_names = ["data/RH DATA 2021-2025.xls", "data/FP 2021-2025.xls"]
for i, file_name in enumerate(file_names):
    df = pd.read_excel(file_name)
    dfs.append(df)

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for i, df in enumerate(dfs):
    dfs[i] = dfs[i].fillna(0)
    dfs[i].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga_id', 'timestamp', 'data_point', 'value']
months = {"JANUARY":"01", "FEBRUARY":"02","MARCH":"03","APRIL":"04","MAY":"05","JUNE":"06","JULY":"07","AUGUST":"08","SEPTEMBER":"09","OCTOBER":"10","NOVEMBER":"11","DECEMBER":"12"}
new_df = pd.DataFrame(columns=columns)
for df in dfs:
    for index, row in df.iterrows():
        lga = " ".join(row["organisationunitname"].upper().split()[1:-3])
        if(len(lga_df.loc[lga_df['lga_name'] == lga, "lga_id"].values) == 0):
            print(lga)

        else:
            for col in df.columns[1:]:
                transformed_data = [{
                    'lga_id': lga_df.loc[lga_df['lga_name'] == lga, "lga_id"].values[0],
                    'timestamp': months[col.upper().split()[-2]] + "-01-" + str(int(col.split()[-1])), #checking if the year is indeed an integer
                    'data_point': " ".join(col.replace("\n", " ").split(" ")[:-2]),
                    'value': row[col]
                }]   

                new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.tail()
```

```python
print("len", len(new_df['data_point'].unique()))
new_df['data_point'].unique() 
```

```python
rh_df = pd.read_excel("data/REPRODUCTIVE HEALTH DATA 2021-2025.xls")
rh_df = rh_df.fillna(0)
rh_df.replace(['NA', 'NIL'], 0, inplace=True)


for index, row in rh_df.iterrows():
    lga = " ".join(row["organisationunitname"].upper().split()[1:-3])
    if(len(lga_df.loc[lga_df['lga_name'] == lga, "lga_id"].values) == 0):
        print(lga)

    else:
        for col in rh_df.columns[2:]:
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == lga, "lga_id"].values[0],
                'timestamp': months[row["periodname"].upper().split()[0]] + "-01-" + str(int(row["periodname"].split()[-1])), #checking if the year is indeed an integer
                'data_point': " ".join(col.replace("\n", " ").split(" ")),
                'value': row[col]
            }]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.tail()
```

```python
print("len", len(new_df['data_point'].unique()))
new_df['data_point'].unique() 
```

```python
new_df["value"]=new_df["value"].astype(int)

new_df.head()
```

```python
new_df.to_excel("transposed/reproductive_health_data.xlsx", index=False)
print("Finished!")
```
