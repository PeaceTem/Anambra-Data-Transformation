```python
import pandas as pd
```

```python
df = pd.read_excel("data/PW and WRA data 2021-2025.xlsx")

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
df = df.fillna(0)
df.replace(['NA', 'NIL'], 0, inplace=True)

df.head()
```

```python
columns = ['lga_id', 'timestamp', 'district_population', 'total_num_of_pw', 'total_no_of_women_reproductive_age']

new_df = pd.DataFrame(columns=columns)

for index, row in df.iterrows():
    if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"].upper(), "lga_id"].values) == 0):
        print(row["LGA"])
    else:
        for i in range(1, len(df.columns), 3):
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"].upper(), "lga_id"].values[0],
                'timestamp': "12-31-" + str(int(df.columns[i].split()[-2])),
                'district_population': round(row[df.columns[i]]),
                # these values are meant to be derived from the district population
                'total_num_of_pw': round(row[df.columns[i+1]]), # 5% of pop
                'total_no_of_women_reproductive_age': round(row[df.columns[i+2]]), # 22% of pop
            }]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['district_population']=new_df['district_population'].astype(int)
new_df['total_num_of_pw']=new_df['total_num_of_pw'].astype(int)
new_df['total_no_of_women_reproductive_age']=new_df['total_no_of_women_reproductive_age'].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposed/pw_and_wra_data.xlsx", index=False)
print("Done!")
```
