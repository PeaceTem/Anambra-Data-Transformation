```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 50", "Table 51", "Table 52", "Table 53"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[1].head()
```

```python
columns = ['lga_id', 'establishment', 'year', 'population']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if(row["LGA"] == "IDIMILI NORTH"): row["LGA"] = "IDEMILI NORTH"
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values) == 0):
            print(row["LGA"])
        # print(row)
        else:
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'establishment': "TEACHING HOSPITAL",
                'year': year,
                'population': row["TEACHING HOSPITAL"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'establishment': "GENERAL HOSPITAL",
                'year': year,
                'population': row["GENERAL HOSPITALS"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'establishment': "PRIMARY HEALTH CENTRE",
                'year': year,
                'population': row["PRIMARY HEALTH CENTRE"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'establishment': "PRIVATE PRIMARY FACILITY",
                'year': year,
                'population': row["PRIVATE PRIMARY FACILITY"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'establishment': "PRIVATE SECONDARY FACILITY",
                'year': year,
                'population': row["PRIVATE SECONDARY FACILITY"],
            },]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['population']=new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("establishments_by_category.xlsx", index=False)
print("Done!")
```
