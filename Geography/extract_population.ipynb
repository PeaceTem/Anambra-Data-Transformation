```python
import pandas as pd
```

```python
year=2022
lga_df = pd.read_csv("Anambra_LGAs.csv")
df = pd.read_excel("Geography & Population.xlsx", sheet_name="T5")
df.head()
```

```python
columns = ['lga_id', 'year', 'gender', 'population']

new_df = pd.DataFrame(columns=columns)


for index, row in df.iterrows():
    
    # if(lga_df.loc[lga_df['lga_name'] == row['LGA'], 'lga_id'].values[0]):
    #     print(row["LGA"], lga_df["lga_name"])
    # else:
    transformed_data = [{
        'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
        'year': year,
        'gender': 'Male',
        'population': row[f"{year} MALE"],
    },{
        'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
        'year': year,
        'gender': 'Female',
        'population': row[f"{year} FEMALE"],
    }]   

    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df["year"]=new_df["year"].astype(int)
new_df["population"]=new_df["population"].astype(int)
new_df["lga_id"]=new_df["lga_id"].astype(int)
```

```python
new_df.to_csv(f"{year}_Anambra_population.csv", index=False)
print(f"{year} Done With this Shit!")
```
