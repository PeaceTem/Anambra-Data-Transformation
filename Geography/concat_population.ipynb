```python
import  pandas as pd
```

```python
df06 = pd.read_csv("2006_Anambra_population.csv")
df19 = pd.read_csv("2019_Anambra_population.csv")
df20 = pd.read_csv("2020_Anambra_population.csv")
df21 = pd.read_csv("2021_Anambra_population.csv")
df22 = pd.read_csv("2022_Anambra_population.csv")

new_df = pd.concat([df06,df19,df20,df21,df22], ignore_index=True)

output_file = "Anambra_population.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    new_df.to_excel(writer, sheet_name="population", index=False)
print("Done Writing!")
```
