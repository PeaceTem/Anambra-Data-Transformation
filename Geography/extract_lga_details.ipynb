```python
import pandas as pd
```

```python
lga_df=pd.read_excel("Anambra_LGAs.xlsx")
df=pd.read_excel("Geography & Population.xlsx", sheet_name="T1")
df.head()
```

```python
columns = ['lga_id', 'lga_name', 'zone', 'AREA_SQ_KM', 'HIGHLANDS_percent', 'PLAIN_LANDS_percent']
new_df = pd.DataFrame(columns=columns)


for index, row in lga_df.iterrows():
    transformed_data = [{
        'lga_id': row['lga_id'],
        'lga_name': row['lga_name'],
        'zone': row['zone'],
        'AREA_SQ_KM': df.loc[df['LGA'] == row["lga_name"], 'AREA_SQ_KM'].values[0],
        'HIGHLANDS_percent': df.loc[df['LGA'] == row["lga_name"], 'HIGHLANDS_percent'].values[0],
        'PLAIN_LANDS_percent': df.loc[df['LGA'] == row["lga_name"], 'PLAIN_LANDS_percent'].values[0],
    }]   

    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df.to_csv("Anambra_LGAs.csv", index=False)
print("Done!")
```
