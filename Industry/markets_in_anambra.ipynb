```python
import pandas as pd
```

```python
file_names = ["data/markets_in_anambra.xlsx"]

dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet1"))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
columns = ['name', 'town', 'lga_id', 'ownership']
df = dfs[0]
df.drop('NO', axis=1, inplace=True)
df.columns = columns
map_values = dict(zip(lga_df['lga_name'], lga_df['lga_id']))
# print(map_values)

df['lga_id'] = df['lga_id'].astype(str).map(lambda x: map_values.get(x.strip().upper())).astype(int)

df.tail(50)
```

```python
df.to_excel("transposedb/markets_in_anambra.xlsx", index=False)
print("Done with transfromation!")
```
