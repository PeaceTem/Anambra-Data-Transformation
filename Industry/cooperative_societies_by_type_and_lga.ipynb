```python
import pandas as pd
```

```python
file_name = ["data/cooperative_societies_by_type_and_lga.xlsx"]
sheets = ['Sheet1', 'Sheet2', 'Sheet3']
years=[2019, 2020, 2021]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel(file_name[0], sheet_name=sheet))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
# In the ways of the old masters
# LordMokuTLE
columns = ['lga_id', 'year', 'type', 'num_of_cooperative_societies']
new_df = pd.DataFrame(columns=columns)
for i, df in enumerate(dfs):
    map_values = dict(zip(lga_df['lga_name'], lga_df['lga_id']))
    # print(map_values)
    df['year']=years[i]
    df.rename(columns={'LGA':'lga_id'}, inplace=True)
    df['lga_id'] = df['lga_id'].astype(str).map(lambda x: map_values.get(x.strip().upper())).astype(int)
    mdf = pd.melt(df, id_vars=['lga_id', 'year'], var_name='type', value_name='num_of_cooperative_societies')
    new_df = pd.concat([new_df, mdf], ignore_index=True)

new_df.head(50)
```

```python
new_df['num_of_cooperative_societies'] = new_df['num_of_cooperative_societies'].astype(int)

new_df.to_excel("transposedb/cooperative_societies_by_type_and_lga.xlsx", index=False)
print("Done with transfromation!")
```
