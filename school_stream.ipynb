```python
import pandas as pd
```

```python
month = "July"
timestamp = "07-01-2024"
df = pd.read_excel("july.xlsx", sheet_name="SCHOOLS SUMMARIES")
school_df = pd.read_csv("combine_schools.csv")
school_df.head()
```

```python
# Reset index for easier manipulation
# df.reset_index(drop=True, inplace=True)
df = df.fillna(0)
df.head(100)
```

```python
lga_df=pd.read_excel("Anambra_LGAs.xlsx")
lga_df.head()
```

```python

# Create a new DataFrame
new_df = pd.DataFrame(columns=['school_id', 'class', 'timestamp', 'stream'])
for index, row in df.iterrows():
    lga_id = lga_df.loc[lga_df['lga_name'] == row['LGA'].strip(), 'lga_id'].values[0]
    school_ids = school_df.loc[(school_df['school_name'] == row['SCHOOL'].strip()) & (lga_id == school_df['lga_id']) & (school_df['category'] == 'Secondary'), 'school_id'].values
    if(len(school_ids) == 0):
        print(row['SCHOOL'].strip(), "is not in the school_df")
    elif(len(school_ids) > 1):
        print(index, school_df.loc[index, 'school_name'], school_ids)
    else:  
        transformed_data = [{
            'school_id': school_ids[0],
            'class': 'JSS1',
            'timestamp': timestamp,
            'stream': row['JSS1_ST'],
        },
        {
            'school_id': school_ids[0],
            'class': 'JSS2',
            'timestamp': timestamp,
            'stream': row['JSS2_ST'],
        },
        {
            'school_id': school_ids[0],
            'class': 'JSS3',
            'timestamp': timestamp,
            'stream': row['JSS3_ST'],
        },
        {
            'school_id': school_ids[0],
            'class': 'SSS1',
            'timestamp': timestamp,
            'stream': row['SSS1_ST'],
        },
        {
            'school_id': school_ids[0],
            'class': 'SSS2',
            'timestamp': timestamp,
            'stream': row['SSS2_ST'],
        },
        {
            'school_id': school_ids[0],
            'class': 'SSS3',
            'timestamp': timestamp,
            'stream': row['SSS3_ST'],
        }]
        # Append the transformed data to the new DataFrame
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)

fg = new_df.loc[500:550]
fg.head(20)
```

```python
new_df.replace(['-', '_', 'NIL', 'O'], 0, inplace=True)
new_df['stream'] = new_df['stream'].astype(int)
new_df['timestamp'] = pd.to_datetime(new_df['timestamp'], format='%m-%d-%Y').dt.strftime('%m-%d-%Y')
new_df.head()
```

```python
new_df.to_csv(f"{month}_school_stream.csv", index=False)
print("Done with this!")
```
