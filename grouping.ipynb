```python
import pandas as pd

# Define the file paths for the CSV files
csv_files_group1 = ["December_school_summary.csv", "January_school_summary.csv", "february_school_summary.csv", "March_school_summary.csv","April_school_summary.csv", "May_school_summary.csv", "June_school_summary.csv", "July_school_summary.csv", "Primary_Schools_Summaries.csv"]  # Group 1: Combined into one sheet
csv_files_group2 = ["December_state_of_origin.csv", "January_state_of_origin.csv", "February_state_of_origin.csv", "march_state_of_origin.csv", "April_state_of_origin.csv", "May_state_of_origin.csv", "june_state_of_origin.csv", "july_state_of_origin.csv"]              # Group 2: Combined into another sheet
csv_files_group3 = ["december_teacher_summary.csv", "january_teacher_summary.csv", "february_teacher_summary.csv", "march_teacher_summary.csv", "April_teacher_summary.csv", "May_teacher_summary.csv", "june_teacher_summary.csv", "july_teacher_summary.csv", "teacher_primary.csv"]  # Group 1: Combined into one sheet
csv_files_group4 = ["December_school_stream.csv", "January_school_stream.csv", "February_school_stream.csv", "March_school_stream.csv","April_school_stream.csv", "May_school_stream.csv", "June_school_stream.csv", "July_school_stream.csv",]  # Group 1: Combined into one sheet


# Combine CSV files in Group 1
df_group1 = pd.concat([pd.read_csv(file) for file in csv_files_group1], ignore_index=True)

# Combine CSV files in Group 2
df_group2 = pd.concat([pd.read_csv(file) for file in csv_files_group2], ignore_index=True)


# Combine CSV files in Group 2
df_group3 = pd.concat([pd.read_csv(file) for file in csv_files_group3], ignore_index=True)


# Combine CSV files in Group 2
df_group4 = pd.concat([pd.read_csv(file) for file in csv_files_group4], ignore_index=True)

# Create an Excel writer object
output_file = "new_school_population.xlsx"
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    # Write Group 1 data to the first sheet
    df_group1.to_excel(writer, sheet_name="SCHOOL SUMMARIES", index=False)
    
    # Write Group 2 data to the second sheet
    df_group2.to_excel(writer, sheet_name="STATE OF ORIGIN", index=False)

    # Write Group 1 data to the first sheet
    df_group3.to_excel(writer, sheet_name="TEACHER SUMMARIES", index=False)

    # Write Group 1 data to the first sheet
    df_group4.to_excel(writer, sheet_name="STREAM SUMMARIES", index=False)


print(f"Data has been successfully written to {output_file}")
```
