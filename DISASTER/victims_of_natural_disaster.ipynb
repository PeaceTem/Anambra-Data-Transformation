```python
import pandas as pd
```

```python
years = [2019, 2020, 2022]
sheets = ["Table 118", "Table 119", "Table 120"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ["lga_id", "year", "type_of_disaster", "no_of_persons_supported",
        "no_of_persons_displaced", "percentage_population_supported",
         "no_of_emergency_centres", "no_of_idp_camps", "no_of_persons_in_idp_camps"]
new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if(row["TYPE OF DISASTER"] in (0,'0')): continue
        transformed_data = [{
            'lga_id': lga_df.loc[lga_df["lga_name"]== row["LGA"], 'lga_id'].values[0],
            'year': year,
            'type_of_disaster': row['TYPE OF DISASTER'],
            'no_of_persons_displaced': row['NUMBER OF PERSONS DISPLACED  BY NATURAL DISASTER'],
            'no_of_persons_supported': row['NUMBER OF PERSONS SUPPORTED'],
            'percentage_population_supported' : row["PERCENTAGE OF THE POPULATION SUPPORTED AFTER DISASTER"],
            'no_of_emergency_centres': row['NUMBER OF OPERATIONAL/ EMERGENCY SUPPORT CENTRE'],
            'no_of_idp_camps': row['NO. OF IDPS CAMPS'],
            'no_of_persons_in_idp_camps': row['NO. OF PERSONS IN IDP CAMPS'],
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['no_of_persons_supported'] = new_df['no_of_persons_supported'].astype(int)
new_df['no_of_persons_displaced'] = new_df['no_of_persons_displaced'].astype(int)
new_df['percentage_population_supported'] = new_df['percentage_population_supported'].astype(int)
new_df['no_of_emergency_centres'] = new_df['no_of_emergency_centres'].astype(int)
new_df['no_of_idp_camps'] = new_df['no_of_idp_camps'].astype(int)
new_df['no_of_persons_in_idp_camps'] = new_df['no_of_persons_in_idp_camps'].astype(int)
new_df.head()
```

```python
new_df.to_excel("victims_of_natural_disaster.xlsx", index=False)
print("Done!")
```
