```python
import pandas as pd
```

```python
years = [2022]
sheets = ["Table 130"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ["year", "premises", "no_of_fire_incidents", "no_of_lives_lost", "no_of_unhurt_persons", "no_of_persons_injured", "est_value_of_property_lost", "est_value_of_property_saved"]
new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        transformed_data = [{
            'year': year,
            'premises': row['DESCRIPTION OF PREMISES'],
            'no_of_fire_incidents': row['NUMBER OF FIRE INCIDENTS'],
            'no_of_lives_lost': row['NUMBER OF LIVES LOST'],
            'no_of_unhurt_persons': row['NUMBER OF PERSONS UNHURT'],
            'no_of_persons_injured': row['NUMBER OF PERSONS INJURED'],
            'est_value_of_property_lost': row['ESTIMATED VALUE OF PROPERTY LOST'],
            'est_value_of_property_saved': row['ESTIMATED VALUE OF PROPERTY SAVED'],
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df.to_excel("fire_incidents_by_type_of_premises.xlsx", index=False)
print("Done!")
```
