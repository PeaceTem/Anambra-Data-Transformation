```python
import pandas as pd
```

```python
years = [2019, 2020, 2022]
sheets = ["Table 115", "Table 116", "Table 117"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

# dfs[1].head()
```

```python
columns = ['id', 'lga_id', 'year', 'no_of_household_affected_by_flood', 'no_of_household_supported']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if row["COMMUNITIES AFFECTED BY FLOOD"] in (0, '0'): continue

        if(row["LGA"] not in (0, '0')): lga=row["LGA"]
        
        if(len(lga_df.loc[lga_df['lga_name'] == lga, "lga_id"]) == 0):
            print(row["LGA"])
        lga_id = lga_df.loc[lga_df["lga_name"] == lga, "lga_id"].values[0]

        transformed_data = [{
            'lga_id': lga_id,
            'year': year,
            'no_of_household_affected_by_flood': row["NUMBER OF HOUSEHOLD AFFECTED BY FLOOD"],
            'no_of_household_supported': row["NUMBER OF HOUSEHOLD SUPPORTED"],
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df['id']=new_df.index
new_df.head(22)
```

```python
new_df["no_of_household_affected_by_flood"] = new_df["no_of_household_affected_by_flood"].astype(int)
new_df["no_of_household_supported"] = new_df["no_of_household_supported"].astype(int)
new_df.head()
```

```python
new_df.to_excel("household_affected_by_flood.xlsx", index=False)
print("Done!")
```

```python
# household_affected_by_flood_id is the id of the household_affected_by_flood table
# communities affected_by_erosion
columns = ['household_affected_by_flood_id', 'community']
community_df = pd.DataFrame(columns=columns)
for i , df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if row["COMMUNITIES AFFECTED BY FLOOD"] in (0, '0'): continue

        if(row["LGA"] not in (0, '0')): lga=row["LGA"]

        if(len(lga_df.loc[lga_df['lga_name'] == lga, "lga_id"]) == 0):
            print(row["LGA"])
        lga_id = lga_df.loc[lga_df["lga_name"] == lga, "lga_id"].values[0]
        communities = row["COMMUNITIES AFFECTED BY FLOOD"].split(',')
        # len_comm = len(new_df.loc[(new_df["lga_id"]==lga_id) & (new_df["year"]==year), "id"].values)
        # if(len_comm>1):
        #     print("hasfklhjsbd")
        # elif(len_comm==0):
        #     print("hagjlaajgkallk")
        # else:
        #     print("working!")

        # print(communities)
        new_df_id = new_df.loc[(new_df["lga_id"]==lga_id) & (new_df["year"]==year), "id"].values[0]
        for community in communities:
            transformed_data = [{
                "household_affected_by_flood_id": new_df_id,
                'community': community.replace("\n", " ").strip()
            },]   

            community_df = pd.concat([community_df, pd.DataFrame(transformed_data)], ignore_index=True)

community_df = community_df.drop_duplicates()
# community_df['id'] = community_df.index
community_df.head(50)
# print(len(community_df))
```
