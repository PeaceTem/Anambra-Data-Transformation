```python
import pandas as pd
```

```python
erosion_sites = pd.read_excel("erosion_sites_in_anambra.xlsx")
erosion_sites.head()
```

```python
lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
mapping = lga_df.set_index('lga_name')['lga_id'].to_dict()
erosion_sites['LGA'] = erosion_sites['LGA'].str.upper().str.strip()
erosion_sites['site'] = erosion_sites['SITE'].str.strip()
erosion_sites['town'] = erosion_sites['TOWN'].str.strip()

erosion_sites["lga_id"] = erosion_sites["LGA"].map(mapping)
erosion_sites = erosion_sites.drop(columns=['LGA', 'SITE', 'TOWN'])

erosion_sites.head(50)
```

```python
erosion_sites.to_excel("transposedb/erosion_sites_in_anambra_transposed.xlsx", index=False)
print("Done!")
```
