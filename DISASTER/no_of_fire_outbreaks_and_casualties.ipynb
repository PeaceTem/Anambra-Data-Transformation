```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 121", "Table 122", "Table 123", "Table 124"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ["lga_id", "year", "no_of_fire_outbreak", "no_of_persons_saved", "no_of_persons_injured", "no_of_death", "est_value_of_property_lost", "est_value_of_property_saved"]
new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        transformed_data = [{
            'lga_id': lga_df.loc[lga_df["lga_name"]== row["LGA"], 'lga_id'].values[0],
            'year': year,
            'no_of_fire_outbreak': row['NUMBER OF FIRE OUTBREAK'],
            'no_of_persons_injured': row['NUMBER OF PERSONS INJURED'],
            'no_of_persons_saved': row['NUMBER OF PERSONS SAVED'],
            'no_of_death' : row["NUMBER OF DEATH"],
            'est_value_of_property_lost': row['ESTIMATED VALUE OF PROPERTIES LOST'],
            'est_value_of_property_saved': row['ESTIMATED VALUE OF PROPERTIES SAVED'],
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df.to_excel("fire_outbreaks_and_casualties.xlsx", index=False)
print("Done!")
```
