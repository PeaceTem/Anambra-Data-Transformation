```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 125", "Table 126", "Table 127", "Table 128"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ["timestamp", "no_of_fire_incidents", "false_call", "no_of_death", "no_of_persons_injured", "est_value_of_property_lost", "est_value_of_property_saved"]
new_df = pd.DataFrame(columns=columns)
months = {"JANUARY":"01", "FEBRUARY":"02","MARCH":"03","APRIL":"04","MAY":"05","JUNE":"06","JULY":"07","AUGUST":"08","SEPTEMBER":"09","OCTOBER":"10","NOVEMBER":"11","DECEMBER":"12"}

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        transformed_data = [{
            'timestamp': months[row["MONTH"]] + "-01-" + str(year),
            'no_of_fire_incidents': row['NUMBER OF FIRE INCIDENTS'],
            'false_call': row["FALSE CALL"],
            'no_of_death' : row["NUMBER OF DEATH"],
            'no_of_persons_injured': row['NUMBER OF INJURED PERSONS'],
            'est_value_of_property_lost': row['ESTIMATED VALUE OF PROPERTIES LOST'],
            'est_value_of_property_saved': row['ESTIMATED VALUE OF PROPERTIES SAVED'],
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df.to_excel("fire_incidents_by_month.xlsx", index=False)
print("Done!")
```
