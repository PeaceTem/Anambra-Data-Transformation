```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 107", "Table 108", "Table 109", "Table 110"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[1].head()
```

```python
columns = ['lga_id', 'year', 'community', 'area_of_land_affected(hectares)', 'no_of_houses_affected', 'area_of_land_reclaimed', 'percentage_completion_of_works']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if row["COMMUNITIES AFFECTED BY EROSION"] in (0, '0'): continue
        if(row["LGA"] not in (0, '0')): lga=row["LGA"]
        
        if(len(lga_df.loc[lga_df['lga_name'] == lga, "lga_id"]) == 0):
            print(row["LGA"])
        # print(row)
        transformed_data = [{
            'lga_id': lga_df.loc[lga_df['lga_name'] == lga, "lga_id"].values[0],
            'year': year,
            'community': row["COMMUNITIES AFFECTED BY EROSION"].replace("\n", " ").strip(),
            'area_of_land_affected(hectares)': row["AREA OF LAND AFFECTED BY EROSION (HECTARES)"],
            'no_of_houses_affected': row["NUMBER OF HOUSES AFFECTED"],
            'area_of_land_reclaimed': row["AREA OF LAND RECLAIMED/STABILIZED"],
            'percentage_completion_of_works': row["PERCENTAGE COMPLETION OF WORKS"]
        },]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.tail(22)
```

```python
new_df["no_of_houses_affected"] = new_df["no_of_houses_affected"].astype(int)
new_df.head()
```

```python
new_df.to_excel("communities_affected_by_erosion.xlsx", index=False)
print("Done!")
```
