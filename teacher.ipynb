```python
import pandas as pd
```

```python
month = "July"
timestamp = "07-01-2024"
df = pd.read_excel("july.xlsx", sheet_name="SCHOOLS SUMMARIES")
school_df = pd.read_csv("combine_schools.csv")
df.head()
```

```python
lga_df = pd.read_excel("Anambra_LGAs.xlsx", sheet_name="Sheet1")
lga_df.head()
```

```python
df = df.fillna(0)

# Create a new DataFrame
new_df = pd.DataFrame(columns=['school_id', 'timestamp', 'gender', 'population'])
for index, row in df.iterrows():
    lga_id = lga_df.loc[lga_df['lga_name'] == row['LGA'].strip(), 'lga_id'].values[0]
    school_ids = school_df.loc[(school_df['school_name'] == row['SCHOOL'].strip()) & (lga_id == school_df['lga_id']) & (school_df['category'] == 'Secondary'), 'school_id'].values
    if(len(school_ids) == 0):
        print(row['SCHOOL'].strip(), "is not in the school_df")
    elif(len(school_ids) > 1):
        print(index, school_df.loc[index, 'school_name'], school_ids)
        # duplicates = df['SCHOOL'][df['SCHOOL'].duplicated(keep=False)].tolist()
        # print(duplicates)
    else: 
        transformed_data = [{
            'school_id': school_df.loc[school_df['school_name'] == row['SCHOOL'].strip(), 'school_id'].values[0],
            'timestamp': timestamp,
            'gender': 'Male',
            'population': row['TEACHER_M'],
        },{
            'school_id': school_df.loc[school_df['school_name'] == row['SCHOOL'].strip(), 'school_id'].values[0],
            'timestamp': timestamp,
            'gender': 'Female',
            'population': row['TEACHER_F'],
        }]
        # Append the transformed data to the new DataFrame
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(200)
```

```python
new_df.replace(['-', '_', 'NIL', 'O', 'Nil'], 0, inplace=True)
new_df['population'] = new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_csv(f"{month}_teacher_summary.csv", index=False)
print("Use LeetCode Design!")
```
