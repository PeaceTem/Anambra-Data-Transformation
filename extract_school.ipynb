```python
import pandas as pd
```

```python
df = pd.read_excel("july.xlsx", sheet_name="STATE OF ORIGIN")
lga_df = pd.read_excel("Anambra_LGAs.xlsx", sheet_name="Sheet1")
df.head()
```

```python
new_df = pd.DataFrame(columns=['lga_id', 'school_id', 'school_name', 'school_type', 'lat', 'long'])
count = 0
for index, row in df.iterrows():
    count += 1
    transformed_data = []
    data = {
    'lga_id': lga_df.loc[lga_df['lga_name'] == row['LGA'], 'lga_id'].values[0],
    'school_id': count,
    'school_name': row['SCHOOL'].strip(),
    'school_type': 'Secondary',
    'lat': '',
    'long': '',
    }
    transformed_data.append(data)
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)

new_df.head()
```

```python
new_df.to_csv("Schools.csv", index=False)
print("Done!")
```
