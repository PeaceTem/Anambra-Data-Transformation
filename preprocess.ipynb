```python
month = "February"
lga = ""

df = pd.read_excel("february.xlsx", sheet_name="SCHOOLS SUMMARIES")
df.head()
```

```python
# Extract the relevant data starting from the correct row
# We'll skip metadata rows and manually set column names
start_row = 0  # Data starts after "S/N" and "NAME OF SCHOOL"
columns = ['s/n', 'school', 'jss1_m', 'jss1_f', 'jss1_t', 'jss1_st',
           'jss2_m', 'jss2_f', 'jss2_t', 'jss2_st', 'jss3_m', 'jss3_f', 
           'jss3_t', 'jss3_st', 'sss1_m', 'sss1_f', 'sss1_t', 'sss1_st', 
           'sss2_m', 'sss2_f', 'sss2_t', 'sss2_st', 'sss3_m', 'sss3_f', 
           'sss3_t', 'sss3_st', 'total_st', 'total_m', 'total_f', 'total_t', 
           'indg', 'non_indg_m', 'non_indg_f', 'non_indg_t']

# Set column names and load relevant rows
data_cleaned = df.iloc[start_row:, :len(columns)]
data_cleaned.columns = columns

# Reset index for easier manipulation
data_cleaned.reset_index(drop=True, inplace=True)
data_cleaned = data_cleaned.fillna(0)
data_cleaned.head(100)
```

```python

# Create a new DataFrame
new_df = pd.DataFrame(columns=['lga', 'school', 'class', 'male', 'female', 'st'])
lga = ''
previous_row = None
for index, row in data_cleaned.iterrows():
    #print(row['s/n'])
    
    # Example transformation: You can customize this as needed
    if row['s/n'] in ['S/N', 'S/NO']: 
        lga = previous_row['school']
        #print(lga)
        continue
    elif isinstance(row['s/n'], int) and row['s/n'] > 0:  
        #print(row['s/n'])
        transformed_data = [{
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'jss1',
            'male': row['jss1_m'],
            'female': row['jss1_f'],
            'st': row['jss1_st'],
        },
                           {
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'jss2',
            'male': row['jss2_m'],
            'female': row['jss2_f'],
            'st': row['jss2_st'],
        },
            {
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'jss3',
            'male': row['jss3_m'],
            'female': row['jss3_f'],
            'st': row['jss3_st'],
        },
            {
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'sss1',
            'male': row['sss1_m'],
            'female': row['sss1_f'],
            'st': row['sss1_st'],
        },
            {
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'sss2',
            'male': row['sss2_m'],
            'female': row['sss2_f'],
            'st': row['sss2_st'],
        },
            {
            'lga': lga,  # Example: Multiply a value by 2
            'school': row['school'],
            'class': 'sss3',
            'male': row['sss3_m'],
            'female': row['sss3_f'],
            'st': row['sss3_st'],
        }]
    previous_row = row
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(200)
```

```python
new_df.to_csv("output.csv", index=True)
print("Done")
```
