```python
import pandas as pd
```

```python
# check if all the schools in school summaries are valid in accordance to the schools in the main secondary schools data.
school_df = pd.read_csv("combine_schools.csv")
df = pd.read_excel("february.xlsx", sheet_name="SCHOOLS SUMMARIES")
```

```python
sec_sch = df['SCHOOL'].tolist()
sch = school_df.loc[school_df['school_type'] != 'PRIVATE', 'school_name'].values
```

```python
s = [x for x in sec_sch if x.strip() in sch]
y =[x for x in sec_sch if x.strip() not in sch]
print(s, y, sep='\n')
```
