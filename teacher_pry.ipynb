```python
import pandas as pd
```

```python
timestamp="01-01-2023"
df= pd.read_excel("primary_school.xlsx")
df.head()
```

```python
df.fillna(0)
school_df = pd.read_csv("combine_schools.csv")
school_df.head()
```

```python
lga_df = pd.read_excel("Anambra_LGAs.xlsx", sheet_name="Sheet1")
lga_df.head()
```

```python
# Create a new DataFrame
new_df = pd.DataFrame(columns=['school_id', 'timestamp', 'gender', 'population'])
for index, row in df.iterrows():
    lga_id = lga_df.loc[lga_df['lga_name'] == row['LGA'].strip(), 'lga_id'].values[0]
    school_ids = school_df.loc[(school_df['school_name'] == row['SCHOOL'].strip()) & (lga_id == school_df['lga_id']) & (school_df['category'] == 'Primary'), 'school_id'].values
    if(len(school_ids) == 0):
        print(row['SCHOOL'].strip(), "is not in the school_df")
    elif(len(school_ids) > 1):
        print(index, school_df.loc[index, 'school_name'], school_ids)
        
    else: 
        transformed_data = [{
            'school_id': school_ids[0],
            'timestamp': timestamp,
            'gender': 'Male',
            'population': row['TEACHER_M'],
        },
        {
            'school_id': school_ids[0],
            'timestamp': timestamp,
            'gender': 'Female',
            'population': row['TEACHER_F'],
        }]
        # Append the transformed data to the new DataFrame
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head()
```

```python
new_df['population'] = new_df['population'].astype(int)
new_df['timestamp'] = pd.to_datetime(new_df['timestamp'], format='%m-%d-%Y').dt.strftime('%m-%d-%Y')
new_df.head()
```

```python
new_df.to_csv("teacher_primary.csv", index=False)
print("Done Printing!")
```
