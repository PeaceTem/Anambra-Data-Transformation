```python
import pandas as pd
```

```python
timestamp="01-07-2023"
df= pd.read_excel("primary_school.xlsx")
df.head()
```

```python
df = df.fillna(0)
```

```python
lga_df = pd.read_excel("Anambra_LGAs.xlsx")
lga_df.head()
```

```python
print(df['SCHOOL_TYPE'].unique().tolist())
```

```python
new_df = pd.DataFrame(columns=['lga_id', 'school_id', 'school_name', 'school_type', 'category', 'lat', 'long'])

count = 2000
for index, row in df.iterrows():
    count += 1
    transformed_data = []
    data = {
    'lga_id': lga_df.loc[lga_df['lga_name'] == row['LGA'], 'lga_id'].values[0],
    'school_id': count,
    'school_name': row['SCHOOL'].strip(),
    'school_type': 'PUBLIC' if row['SCHOOL_TYPE'] == 0 else 'PUBLIC MISSION' if any(x in row['SCHOOL_TYPE'] for x in ('CATHOLIC', 'ANGLICAN')) else 'PUBLIC' if any(x in row['SCHOOL_TYPE'] for x in ('PUBLIC')) else 'PUBLIC',
    'category': 'Primary',
    'lat': '',
    'long': '',
    }
    transformed_data.append(data)
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)
```

```python
new_df.head(30)
```

```python
new_df.to_csv("Primary_Schools.csv", index=False)
print("Done Printing!")
```
