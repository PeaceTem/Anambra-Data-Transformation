```python
import pandas as pd
```

```python
years = [2022]
sheets = ["Table 186"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['timestamp', 'airlines', 'flights', 'arriving_passengers', 'departing_passengers']
months = {"JANUARY":"01", "FEBRUARY":"02","MARCH":"03","APRIL":"04","MAY":"05","JUNE":"06","JULY":"07","AUGUST":"08","SEPTEMBER":"09","OCTOBER":"10","NOVEMBER":"11","DECEMBER":"12"}


new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    df["MONTH"]=df["MONTH"].astype(str)
    for i, row in df.iterrows():
        if(row["MONTH"]=="TOTAL"): continue
        if(row["MONTH"]!="0"): month=row["MONTH"]
        
        transformed_data = [{
            'timestamp': str(months[month])+"-01-"+str(year),
            'airlines': row["AIRLINES"],
            'flights': row["NUMBER OF FLIGHTS (ARRIVALS & DEPARTURE)"],
            'arriving_passengers': row["NUMBER OF ARRIVING PASSENGERS"],
            'departing_passengers': row["NUMBER OF DEPARTING PASSENGERS"],
        }]
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(50)
```

```python
new_df["flights"]=new_df["flights"].astype(int)
new_df["arriving_passengers"]=new_df["arriving_passengers"].astype(int)
new_df["departing_passengers"]=new_df["departing_passengers"].astype(int)

new_df.to_excel("airlines_traffic.xlsx", index=False)
print("Done!")
```
