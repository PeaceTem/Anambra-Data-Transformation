```python
import pandas as pd
```

```python
education_levels = ["Secondary", "Primary"]
filenames = ["data/CORE PUBLIC SECONDARY SCHOOLS.xlsx", "data/CORE PUBLIC PRIMARY SCHOOLS.xlsx"]
dfs = []
for index, filename in enumerate(filenames):
    dfs.append(pd.read_excel(filename))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[0].head()
```

```python
columns = ['lga_id', 'session', 'education_level', 'gender', 'population']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGEA"], "lga_id"].values) == 0):
            print(row["LGEA"])
        # print(row)
        else:
            for col in df.columns[2:]:
                transformed_data = [{
                    'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGEA"], "lga_id"].values[0],
                    'session': col.strip().split('_')[1],
                    'education_level': education_levels[i],
                    'gender': "Male" if col.strip().split('_')[0] == 'M' else "FEMALE",
                    'population': row[col],
                }]   

                new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['population']=new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposedb/core_public_schools_student_population.xlsx", index=False)
print("Done!")
```
