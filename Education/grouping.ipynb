```python
import pandas as pd
```

```python
# Define the file paths for the CSV files
csv_files_group1 = ["Pre_Primary_School_Population.xlsx", "Primary_School_Population.xlsx", "Junior_Secondary_School_Population.xlsx", "Senior_Secondary_School_Population.xlsx", "2024_Secondary_Schools_Population.xlsx"]  # Group 1: Combined into one sheet
csv_files_group2 = ["Primary_School_Population_By_Age_Group.xlsx", "Secondary_School_Population_By_Age_Group.xlsx"]              # Group 2: Combined into another sheet
csv_files_group3 = ["primary_school_teacher_population.xlsx", "secondary_school_teacher_population.xlsx"]              # Group 2: Combined into another sheet


# Combine CSV files in Group 1
df_group1 = pd.concat([pd.read_excel(file) for file in csv_files_group1], ignore_index=True)

# Combine CSV files in Group 2
df_group2 = pd.concat([pd.read_excel(file) for file in csv_files_group2], ignore_index=True)


# Combine CSV files in Group 2
df_group3 = pd.concat([pd.read_excel(file) for file in csv_files_group3], ignore_index=True)

# Create an Excel writer object
output_file = "new_school_population.xlsx"
with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    # Write Group 1 data to the first sheet
    df_group1.to_excel(writer, sheet_name="SCHOOL POPULATION", index=False)
    
    # Write Group 2 data to the second sheet
    df_group2.to_excel(writer, sheet_name="POPULATION BY AGE", index=False)

    # Write Group 1 data to the first sheet
    df_group3.to_excel(writer, sheet_name="TEACHER POPULATION", index=False)

print(f"Data has been successfully written to {output_file}")
```
