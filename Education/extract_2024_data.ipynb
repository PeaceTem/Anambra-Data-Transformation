```python
import pandas as pd
```

```python
# take the last month for both primary and secondary schools
session = "2023/2024"
primary_df = pd.read_csv("combine_schools.csv")
july_school_summaries = pd.read_csv("July_school_summary.csv")
july_school_summaries.head(10)
```

```python
primary_df.head()
```

```python
columns = ['lga_id', 'session', 'class', 'gender', 'population']

new_df = pd.DataFrame(columns=columns)

for index, row in july_school_summaries.iterrows():
    if(len(primary_df.loc[primary_df['school_id'] == row["school_id"], "lga_id"].values) == 0):
        print(row["school_id"])
    # print(row)
    else:
        transformed_data = [{
            'lga_id': primary_df.loc[primary_df['school_id'] == row["school_id"], "lga_id"].values[0],
            'session': session,
            'class': row['class'][:-1] + ' ' + row['class'][-1],
            'gender': row['gender'],
            'population': row["population"],
        }]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(50)
```

```python
new_df['population'] = new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("2024_Secondary_Schools_Population.xlsx", index=False)
print("!Done")
```
