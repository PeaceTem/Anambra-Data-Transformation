```python
school_df=pd.read_csv("combine_schools.csv")
school_df.head()
```

```python
l=[]
for index, row in july_df.iterrows():
    l.append(school_df.loc[school_df["school_id"]==row["school_id"], "lga_id"].values[0])

july_df["lga_id"] = l
july_df.head(50)
```

```python
import numpy as np
for i in range(10,22):
    filtered_df = july_df[(july_df["lga_id"]==i) & (july_df["state"]=="FOREIGN")]
    print(i, len(filtered_df))
    print(np.sum(filtered_df["population"]))
```
