```python
import pandas as pd
```

```python
# take the last month for both primary and secondary schools
session = "2023/2024"
sch_df = pd.read_csv("combine_schools.csv")
july_state_of_origin = pd.read_csv("../july_state_of_origin.csv")
july_state_of_origin.head(10)
# july_state_of_origin.dtypes
```

```python
l=[]
for index, row in july_state_of_origin.iterrows():
    l.append(sch_df.loc[sch_df["school_id"]==row["school_id"], "lga_id"].values[0])

july_state_of_origin["lga_id"] = l
july_state_of_origin.head(50)
```

```python
sch_df.head()
```

```python
import numpy as np
columns = ['lga_id', 'session', 'education_level', 'state', 'population']

new_df = pd.DataFrame(columns=columns)
july_state_of_origin["population"] = july_state_of_origin["population"].astype(int)
new_df["population"] = new_df["population"].astype(int)
for index, row in july_state_of_origin.iterrows():
    lga_id = sch_df.loc[sch_df['school_id'] == row["school_id"], "lga_id"].values[0]
    if(len(new_df[(new_df["lga_id"]==lga_id)&(new_df["state"]==row["state"])])==0):
        filtered_df = july_state_of_origin[(july_state_of_origin["lga_id"]==row["lga_id"]) & (july_state_of_origin["state"]==row["state"])]
        transformed_data = [{
            'lga_id': lga_id,
            'session': session,
            'education_level': "Secondary",
            'state': row['state'],
            'population': np.sum(filtered_df["population"]),
        }]   

        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(50)
```

```python
new_df.to_excel("Secondary_State_of_Origin.xlsx", index=False)
print("Done!")
```
