```python
import pandas as pd
```

```python
df= pd.read_excel("secondary_schools.xlsx", sheet_name="STATE SECONDARY SHOOLS-Grid vie")
df.tail()
```

```python
lga_df = pd.read_excel("Anambra_LGAs.xlsx")
lga_df.head()
```

```python
new_df = pd.DataFrame(columns=['lga_id', 'school_id', 'school_name', 'school_type', 'category', 'lat', 'long'])

count = 0
for index, row in df.iterrows():
    count += 1
    transformed_data = []
    data = {
    'lga_id': lga_df.loc[lga_df['lga_name'] == row['LGA'], 'lga_id'].values[0],
    'school_id': count,
    'school_name': row['NAME OF SCHOOL'].strip(),
    'school_type': 'PUBLIC' if 'GOVERNMENT' in row['REMARKS'] else 'PRIVATE' if 'PRIVATE' in row['REMARKS'] else 'UNKNOWN',
    'category': "Secondary",
    'lat': '',
    'long': '',
    }
    transformed_data.append(data)
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)

new_df.head()
```

```python
print(new_df['school_type'].unique().tolist())
```

```python
new_df.to_csv("Secondary_Schools.csv", index=False)
print("Done Printing!")
```
