```python
import pandas as pd
```

```python
month = "July"
df = pd.read_excel("july.xlsx", sheet_name="STATE OF ORIGIN")
df.head()
```

```python
df = df.fillna(0)
df.head(100)
```

```python

# Create a new DataFrame
new_df = pd.DataFrame(columns=['lga', 'school', 'state', 'month', 'population'])
for index, row in df.iterrows():
    transformed_data = []
    for column in df.columns[3:-4]:
        data = {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'state': column,
        'month': month,
        'population': row[column],
        }
        transformed_data.append(data)
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(200)
```

```python
new_df['population'] = new_df['population'].astype(int)
```

```python
new_df.to_csv("{month}_state_of_origin.csv", index=False)
print("Done")
```
