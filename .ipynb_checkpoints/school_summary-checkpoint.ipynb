```python
import pandas as pd
```

```python
month = "June"
df = pd.read_excel("june.xlsx", sheet_name="SCHOOLS SUMMARIES")
df.head()
```

```python
# Reset index for easier manipulation
# df.reset_index(drop=True, inplace=True)
df = df.fillna(0)
df.head(100)
```

```python

# Create a new DataFrame
new_df = pd.DataFrame(columns=['lga', 'school', 'class', 'month', 'male', 'female', 'stream'])
for index, row in df.iterrows():
    transformed_data = [{
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'JSS1',
        'month': month,
        'male': row['JSS1_M'],
        'female': row['JSS1_F'],
        'stream': row['JSS1_ST'],
    },
    {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'JSS2',
        'month': month,
        'male': row['JSS2_M'],
        'female': row['JSS2_F'],
        'stream': row['JSS2_ST'],
    },
    {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'JSS3',
        'month': month,
        'male': row['JSS3_M'],
        'female': row['JSS3_F'],
        'stream': row['JSS3_ST'],
    },
    {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'SSS1',
        'month': month,
        'male': row['SSS1_M'],
        'female': row['SSS1_F'],
        'stream': row['SSS1_ST'],
    },
    {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'SSS2',
        'month': month,
        'male': row['SSS2_M'],
        'female': row['SSS2_F'],
        'stream': row['SSS2_ST'],
    },
    {
        'lga': row['LGA'],
        'school': row['SCHOOL'],
        'class': 'SSS3',
        'month': month,
        'male': row['SSS3_M'],
        'female': row['SSS3_F'],
        'stream': row['SSS3_ST'],
    }]
    # Append the transformed data to the new DataFrame
    new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(200)
```

```python
new_df['male'] = new_df['male'].astype(int)
new_df['female'] = new_df['female'].astype(int)
new_df['stream'] = new_df['stream'].astype(int)
new_df.head()
```

```python
new_df.to_csv(f"{month}_school_summary.csv", index=False)
print("Done!")
```
