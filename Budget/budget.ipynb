```python
import pandas as pd
pd.options.display.float_format = '{:.2f}'.format 
```

```python
sheets = ["Sheet1"]
file_names = ["./data/budget_2022.xlsx", "./data/budget_2023.xlsx", "./data/budget_2024.xlsx", "./data/budget_2025.xlsx"]
years = [2022, 2023, 2024, 2025]
dfs = []
for index, file_name in enumerate(file_names):
    dfs.append(pd.read_excel(file_name, sheet_name="Sheet1"))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL', '-'], 0, inplace=True)

dfs[0].head()
```

```python
# columns = ['Ministry', 'Personnel Cost', 'Overhead Cost', 'response']

new_df = pd.DataFrame(columns=["year", *dfs[0].columns])

for i, df in enumerate(dfs):
    year = years[i]
    df["year"] = year
    
    new_df = pd.concat([new_df, df], ignore_index=True)


new_df.head(50)
```

```python
print(len(new_df))
```

```python
new_df.to_excel("transposed/budget.xlsx", index=False)
print("Finished!")
```
