```python
import pandas as pd
```

```python
df = pd.read_excel("no_of_public_building.xlsx")
```

```python
df = df.fillna(0)
df.replace(['NA', 'NIL'], 0, inplace=True)
df.replace('s', 5, inplace=True)
```

```python
lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
columns = ['lga_id', 'timestamp_year', 'num_of_buildings']

new_df = pd.DataFrame(columns=columns)

for i, row in df.iterrows():   
    lga_id =  lga_df.loc[lga_df["lga_name"]==row["LGA"].strip().upper(), 'lga_id'].values[0]
    
    for column in df.columns[1:]:
        
        transformed_data = [{
            'lga_id': lga_id,
            'timestamp_year': "01-01-"+str(column),
            'num_of_buildings': row[column],
        }]
        new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df["num_of_buildings"] = new_df["num_of_buildings"].astype(int)
new_df.head()
```

```python
new_df.to_excel("transposedb/no_of_public_buildings.xlsx", index=False)
print("Done!")
```
