```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 103", "Table 104", "Table 105", "Table 106"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("../AGRICULTURE/anambra_2022_year_book.xlsx", sheet_name=sheets[index]))
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)
    dfs[index].replace('s', 5, inplace=True)


dfs[1].head()
```

```python
lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
columns = ['lga_id', 'timestamp', 'building_plans']
months = {"JAN":"01", "FEB":"02","MAR":"03","APRIL":"04","MAY":"05","JUNE":"06","JULY":"07","AUG":"08","SEPT":"09","OCT":"10","NOV":"11","DEC":"12"}


new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for i, row in df.iterrows():
        lgas = row["LGA"].split('/')
        print(lgas)
        for lga in lgas:
            for column in df.columns[2:-1]:
                
                transformed_data = [{
                    'lga_id': lga_df.loc[lga_df["lga_name"]==(lga if lga != 'WEST' else 'ANAMBRA WEST'), 'lga_id'].values[0],
                    'timestamp': str(months[column])+"-01-"+str(year),
                    'building_plans': row[column],
                }]
                new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df["building_plans"] = new_df["building_plans"].astype(int)
new_df.head()
```

```python
new_df.to_excel("building_plan.xlsx", index=False)
print("Done!")
```
