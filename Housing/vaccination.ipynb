```python
import pandas as pd
```

```python
years = [2019, 2020, 2021, 2022]
sheets = ["Table 14", "Table 15", "Table 16", "Table 17"]
dfs = []
for index, sheet in enumerate(sheets):
    dfs.append(pd.read_excel("anambra_2022_year_book.xlsx", sheet_name=sheets[index]))

lga_df = pd.read_csv("../Geography/Anambra_LGAs.csv")
lga_df.head()
```

```python
for index, df in enumerate(dfs):
    dfs[index] = dfs[index].fillna(0)
    dfs[index].replace(['NA', 'NIL'], 0, inplace=True)

dfs[1].head()
```

```python
columns = ['lga_id', 'livestock_slaughtered', 'year', 'population']

new_df = pd.DataFrame(columns=columns)

for i, df in enumerate(dfs):
    year = years[i]
    for index, row in df.iterrows():
        if(len(lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values) == 0):
            print(row["LGA"])
        # print(row)
        else:
            transformed_data = [{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'livestock_slaughtered': "CATTLE",
                'year': year,
                'population': row["CATTLE"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'livestock_slaughtered': "SHEEP",
                'year': year,
                'population': row["SHEEP"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'livestock_slaughtered': "GOAT",
                'year': year,
                'population': row["GOAT"],
            },{
                'lga_id': lga_df.loc[lga_df['lga_name'] == row["LGA"], "lga_id"].values[0],
                'livestock_slaughtered': "PIG",
                'year': year,
                'population': row["PIG"],
            },]   

            new_df = pd.concat([new_df, pd.DataFrame(transformed_data)], ignore_index=True)


new_df.head(22)
```

```python
new_df['population']=new_df['population'].astype(int)
new_df.head()
```

```python
new_df.to_excel("slaughtered_livestock.xlsx", index=False)
print("Done!")
```
